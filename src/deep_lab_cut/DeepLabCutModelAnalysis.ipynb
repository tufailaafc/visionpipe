Visualizing the error in pixels for all parts plotted against their confidence

```python
import numpy as np
import pandas as pd
from deeplabcut.utils import auxiliaryfunctions
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
cfg = auxiliaryfunctions.read_config(project_path)
scorer = auxiliaryfunctions.get_scorer_name(
    cfg,
    shuffle=1,
    trainFraction=cfg["TrainingFraction"][0],
    snapshot="best"
)
```

```python
gt_file = Path(project_root_path) / "training-datasets" / "iteration-0" / \
          f"UnaugmentedDataSet_{cfg['Task']}{cfg['date']}" / \
          f"CollectedData_{cfg['scorer']}.csv"

# pred_file = Path(project_root_path) / "evaluation-results-pytorch" / \
#             f"DLC_Resnet50_{cfg['Task']}{cfg['date']}shuffle1_snapshot-best-440.h5"
iteration = 0
shuffle = 1
trainset = int(cfg["TrainingFraction"][0] * 100)  # 95
snapshot = "best-440"  # or "best-440", "500", etc.

pred_file = (
    Path(project_root_path)
    / "evaluation-results-pytorch"
    / f"iteration-{iteration}"
    / f"{cfg['Task']}{cfg['date']}-trainset{trainset}shuffle{shuffle}"
    / f"DLC_Resnet50_{cfg['Task']}{cfg['date']}shuffle{shuffle}_snapshot_{snapshot}.h5"
)
```

```python
gt = pd.read_csv(gt_file, header=[0,1,2], index_col=0)
pred = pd.read_hdf(pred_file)
```

```python
print("Prediction column levels:")
for i, lvl in enumerate(pred.columns.levels):
    print(f"Level {i}:", list(lvl))
```

```python
# Bodyparts must come from prediction level 2
bodyparts = pred.columns.get_level_values(2).unique()
individual = "animal"  # from Level 1

records = []

for bp in bodyparts:
    # --- Ground truth (3-level) ---
    gt_xy = gt.loc[:, (slice(None), bp, ["x", "y"])].values

    # --- Predictions (4-level) ---
    pr_xy = pred.loc[:, (slice(None), individual, bp, ["x", "y"])].values
    lk = pred.loc[:, (slice(None), individual, bp, "likelihood")].values.squeeze()

    # Valid GT rows
    valid = ~np.isnan(gt_xy).any(axis=1)

    dist = np.linalg.norm(gt_xy[valid] - pr_xy[valid], axis=1)
    lk_valid = lk[valid]

    for d, l in zip(dist, lk_valid):
        records.append({
            "bodypart": bp,
            "error_px": d,
            "likelihood": l
        })
```

```python
df = pd.DataFrame(records)
```

```python

for bp in df["bodypart"].unique():
    sub = df[df["bodypart"] == bp]

    plt.figure(figsize=(5,4))
    plt.scatter(sub["likelihood"], sub["error_px"], s=6, alpha=0.4)
    plt.axvline(0.1, color="red", linestyle="--", label="pcutoff=0.1")
    plt.xlabel("Likelihood")
    plt.ylabel("Pixel Error")
    plt.title(bp)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
summary = (
    df[df["likelihood"] < 0.1]
    .groupby("bodypart")["error_px"]
    .median()
    .sort_values()
)

summary
```
